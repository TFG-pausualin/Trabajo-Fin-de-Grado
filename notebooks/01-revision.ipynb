```python
ids_resenas = set(muestra_resenas["parent_asin"].dropna())
ids_metadatos = set(muestra_metadatos["parent_asin"].dropna())

ids_comunes = ids_resenas.intersection(ids_metadatos)

print("Identificadores comunes:", len(ids_comunes))
print(
    "Porcentaje de productos de la muestra de reseñas con metadatos:",
    len(ids_comunes) / len(ids_resenas) * 100
)
```

```python
ids_resenas = set(
    muestra_resenas["parent_asin"]
    .dropna()
    .astype(str)
    .unique()
)

metadatos_encontrados = []

for bloque in pd.read_json(
    ruta_metadatos,
    lines=True,
    compression="gzip",
    chunksize=100_000
):
    coincidencias = bloque[
        bloque["parent_asin"].astype(str).isin(ids_resenas)
    ]

    if not coincidencias.empty:
        metadatos_encontrados.append(coincidencias)

if metadatos_encontrados:
    metadatos_muestra_resenas = pd.concat(
        metadatos_encontrados,
        ignore_index=True
    )
else:
    metadatos_muestra_resenas = pd.DataFrame()

print(
    "Productos buscados:",
    len(ids_resenas)
)

print(
    "Productos con metadatos encontrados:",
    metadatos_muestra_resenas["parent_asin"].nunique()
)

porcentaje = (
    metadatos_muestra_resenas["parent_asin"].nunique()
    / len(ids_resenas)
    * 100
)

print(
    "Porcentaje de productos con metadatos:",
    round(porcentaje, 2),
    "%"
)
```

```python
print(
    "Filas de metadatos encontradas:",
    len(metadatos_muestra_resenas)
)

print(
    "Productos únicos:",
    metadatos_muestra_resenas["parent_asin"].nunique()
)

print(
    "Duplicados de parent_asin:",
    metadatos_muestra_resenas["parent_asin"].duplicated().sum()
)
```

```python
columnas_metadatos = [
    "parent_asin",
    "main_category",
    "title",
    "average_rating",
    "rating_number",
    "store",
    "categories"
]

metadatos_reducidos = metadatos_muestra_resenas[
    columnas_metadatos
].rename(
    columns={"title": "product_title"}
)

muestra_combinada = muestra_resenas.merge(
    metadatos_reducidos,
    on="parent_asin",
    how="left",
    validate="many_to_one"
)

print("Filas antes de la unión:", len(muestra_resenas))
print("Filas después de la unión:", len(muestra_combinada))
print(
    "Reseñas sin metadatos:",
    muestra_combinada["product_title"].isna().sum()
)

muestra_combinada.head()
```

```python
def es_videojuego(categorias):
    return (
        isinstance(categorias, list)
        and "Games" in categorias
    )

muestra_combinada["es_videojuego"] = (
    muestra_combinada["categories"].apply(es_videojuego)
)

total = len(muestra_combinada)
videojuegos = muestra_combinada["es_videojuego"].sum()
otros_productos = total - videojuegos

porcentaje_videojuegos = videojuegos / total * 100
porcentaje_otros = otros_productos / total * 100

print("RESULTADO DEL FILTRADO")
print("----------------------")
print(f"Total de reseñas analizadas: {total}")
print(f"Reseñas de videojuegos: {videojuegos} ({porcentaje_videojuegos:.1f} %)")
print(f"Reseñas de otros productos: {otros_productos} ({porcentaje_otros:.1f} %)")
```
